# 🗣️ TechnoCrazy — Clonador de Voz Rafael (XTTS v2)
**Clona tu voz con 10 segundos de audio → narra cualquier texto con tu voz**

### Instrucciones:
1. Runtime → Change runtime type → **T4 GPU**
2. Ejecuta cada celda en orden

4. Pega el script de Claude
5. Descarga el audio generado con tu voz

```
# CELDA 1 — Instalar TTS (XTTS v2)
!pip install -q TTS
print('✅ TTS instalado')
```

```
# CELDA 2 — Cargar modelo XTTS v2 (solo primera vez, ~3 min)
from TTS.api import TTS
import torch

device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f'Usando: {device.upper()}')

tts = TTS('tts_models/multilingual/multi-dataset/xtts_v2').to(device)
print('✅ Modelo XTTS v2 listo')
```

```
# CELDA 3 — Subir AUDIO DE REFERENCIA de Rafael
from google.colab import files
import shutil

print('🎙️ Sube un audio tuyo de 10-30 segundos:')
print('   - Habla claro y natural')
print('   - Sin música ni ruido de fondo')
print('   - MP3 o WAV')

uploaded = files.upload()
ref_audio = list(uploaded.keys())[0]
print(f'\n✅ Audio de referencia cargado: {ref_audio}')
```

```
# CELDA 4 — PEGAR EL SCRIPT Y GENERAR AUDIO
# ⬇️ CAMBIA ESTE TEXTO POR EL SCRIPT GENERADO POR CLAUDE

SCRIPT = """
¿Todavía estás publicando contenido a mano en redes sociales?
Eso te está costando horas que podrías usar para crecer tu negocio.
En TechnoCrazy creamos sistemas que publican solos, con tu cara y tu voz,
mientras tú te enfocas en lo que importa.
Entra a technocrazy.org y conoce el Sistema Avatar IA.
"""

output_path = '/content/audio_rafael.wav'

print('🎙️ Generando audio con tu voz clonada...')
tts.tts_to_file(
    text=SCRIPT.strip(),
    speaker_wav=ref_audio,
    language='es',
    file_path=output_path
)
print(f'✅ Audio generado: {output_path}')
```

```
# CELDA 5 — ESCUCHAR Y DESCARGAR
from IPython.display import Audio, display
from google.colab import files

print('🎧 Reproduciendo el audio generado:')
display(Audio(output_path))

print('\n⬇️ Descargando...')
files.download(output_path)
print('\n✅ Descargado. Ahora sube este audio al notebook de LivePortrait para generar el video.')
```

```
# CELDA 6 — MODO BATCH: Generar varios audios de una vez (para la semana completa)
from google.colab import drive
import os
from datetime import datetime

# Scripts de la semana (pegar los 7 que generó Claude)
scripts_semana = [
    {"dia": "lunes",    "tipo": "viral",  "texto": "PEGA AQUÍ EL SCRIPT DEL LUNES"},
    {"dia": "martes",   "tipo": "edu",    "texto": "PEGA AQUÍ EL SCRIPT DEL MARTES"},
    {"dia": "miercoles","tipo": "viral",  "texto": "PEGA AQUÍ EL SCRIPT DEL MIÉRCOLES"},
    {"dia": "jueves",   "tipo": "venta",  "texto": "PEGA AQUÍ EL SCRIPT DEL JUEVES"},
    {"dia": "viernes",  "tipo": "insp",   "texto": "PEGA AQUÍ EL SCRIPT DEL VIERNES"},
    {"dia": "sabado",   "tipo": "btc",    "texto": "PEGA AQUÍ EL SCRIPT DEL SÁBADO"},
    {"dia": "domingo",  "tipo": "venta",  "texto": "PEGA AQUÍ EL SCRIPT DEL DOMINGO"},
]

drive.mount('/content/drive')
carpeta = '/content/drive/MyDrive/TechnoCrazy/Contenido/Audios'
os.makedirs(carpeta, exist_ok=True)
fecha = datetime.now().strftime('%Y-W%V')

for s in scripts_semana:
    if 'PEGA AQUÍ' not in s['texto']:
        out = f"{carpeta}/TC_{fecha}_{s['dia']}.wav"
        tts.tts_to_file(text=s['texto'], speaker_wav=ref_audio, language='es', file_path=out)
        print(f"✅ {s['dia'].capitalize()}: {out}")

print('\n🎉 Todos los audios generados y guardados en Drive')
```
